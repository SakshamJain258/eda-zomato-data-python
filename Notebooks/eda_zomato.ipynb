# Exploratory Data Analysis (EDA) – Dataset Overview

The dataset contains **51,717 restaurant records and 12 features**, representing various attributes related to restaurant services, ratings, costs, and locations.

After the data cleaning process, **no missing values are present** in the dataset, making it suitable for further exploratory analysis and machine learning tasks.

---

## Feature Classification

The features in the dataset can be categorized into numerical, categorical, and mixed-type variables as follows:

## Numerical Features
- **Rating**  
- **Weighted_Rating**  
- **No._of_rating**  
- **No._of_rating_log**  
- **Avg._cost(2 person)**  
- **Avg_cost(log)**  

---

### Categorical Features
- **online_order** (binary)  
- **Table_booking** (binary)  
- **rest_type**  
- **Listed_type**  
- **Listed_city**  

---

### Mixed-Type Feature
- **cuisines** – contains multiple cuisine values in a single record (e.g., *North Indian, Chinese*)

---
